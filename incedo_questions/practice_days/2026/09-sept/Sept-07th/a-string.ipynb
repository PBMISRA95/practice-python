1. Reverse a string<br>
Write a function that takes a string as an input and returns it reversed.<br>
Example: reverse_string("hello") -> "olleh"

```python
input_str1 = "hello"
def reverse_string(input_str: str) -> str:
    return input_str[::-1]

print(f"The reverse of {input_str1} is : {reverse_string(input_str1)}")
```

2. Count Vowels<br>
Write a function that counts the number of vowels (a, e, i, o, u) in a string.<br>
 Example: count_vowels("hello") -> 2

```python
input_str2 = "hello"
def count_vowels(input_str: str) -> int:
    vowels_list = ['a', 'e', 'i', 'o', 'u']
    vowel_count = 0
    for char in input_str:
        if char in vowels_list:
            vowel_count += 1
    return vowel_count

print(f"The vowel count for {input_str2} is {count_vowels(input_str2)}")
```

3. Palindrome Check<br>
Write a function that checks if a string is a palindrome (case-sensitive)<br>
Example: is_palindrome("Racecar") -> True

```python
input_str3 = "RacecaR"

def is_palindrome(input_str: str) -> bool:
    lower_input_str = input_str.lower()
    left_pointer = 0
    right_pointer = len(input_str) -1
    while left_pointer < right_pointer:
        if lower_input_str[left_pointer] != lower_input_str[right_pointer]:
            return False
        else:
            left_pointer += 1
            right_pointer -= 1
    return True
print(f"{input_str3} is Palindrome : {is_palindrome(input_str3)}")
```

5. Character Frequency<br>
Write a function that returns a dictionary containing the frequency of each character in a string.<br>
Example: count_chars("hello") -> {'h': 1, 'e': 1, 'l': 2, 'o': 1}

```python
input_str5 = "hello"

def count_chars(input_str: str) -> dict:
    freq = {}
    for char in input_str:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(f"The frequency of each string in {input_str} is {count_chars(input_str5)}")
```
